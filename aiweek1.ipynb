```python
board = [['-' for _ in range(3)] for _ in range(3)]
print(board)
turn = 1

while True:
    if turn == 1:
        print("Player1 turn\n")
        player_value = 0
    else:
        print("player2 turn\n")
        player_value = 1

    row = int(input("Enter your row(0,1,2): "))
    col = int(input("Enter your column(0,1,2): "))

    while board[row][col] != '-':
        print("Already occupied, choose another")
        row = int(input("Enter your row(0,1,2): "))
        col = int(input("Enter your column(0,1,2): "))

    board[row][col] = player_value
    print(board)

    game_won = False

    if not game_won:
      for i in range(3):
          if board[i][0] == board[i][1] == board[i][2] == 0:
              print("Player 1 wins!")
              game_won = True
              break
          elif board[i][0] == board[i][1] == board[i][2] == 1:
              print("Player 2 wins!")
              game_won = True
              break

    if not game_won:
        for j in range(3):
            if board[0][j] == board[1][j] == board[2][j] == 0:
                print("Player 1 wins!")
                game_won = True
                break
            elif board[0][j] == board[1][j] == board[2][j] == 1:
                print("Player 2 wins!")
                game_won = True
                break

    if not game_won:
        if board[0][0] == board[1][1] == board[2][2] == 0 or board[0][2] == board[1][1] == board[2][0] == 0:
            print("Player 1 wins!")
            game_won = True
        elif board[0][0] == board[1][1] == board[2][2] == 1 or board[0][2] == board[1][1] == board[2][0] == 1:
            print("Player 2 wins!")
            game_won = True

    if game_won:
        break

    is_draw = True
    for row_list in board:
        if '-' in row_list:
            is_draw = False
            break

    if is_draw:
        print("It's a draw!")
        break


    if turn == 1:
        turn = 2
    else:
        turn = 1
```
