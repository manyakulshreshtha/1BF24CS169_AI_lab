```python
def read_state(name):
    print(f"\nEnter {name} state:")

    state = []

    for i in range(9):
        while True:
            try:
                value = int(input(f"Enter element {i + 1}: "))

                if value < 0 or value > 8:
                    print("Enter a number from 0 to 8.")
                    continue

                if value in state:
                    print("That number is already entered.")
                    continue

                state.append(value)
                break

            except ValueError:
                print("Please enter an integer.")

    return tuple(state)


def display(state):
    print()
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def get_neighbors(state):
    neighbors = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = (
                new_state[new_blank],
                new_state[blank]
            )

            neighbors.append(tuple(new_state))

    return neighbors


def inversion_count(state):
    values = [x for x in state if x != 0]

    count = 0

    for i in range(len(values)):
        for j in range(i + 1, len(values)):
            if values[i] > values[j]:
                count += 1

    return count


def is_solvable(start, goal):
    return inversion_count(start) % 2 == inversion_count(goal) % 2


def reconstruct_path(parent, goal):
    path = []

    current = goal

    while current is not None:
        path.append(current)
        current = parent[current]

    path.reverse()

    return path


def dfs(start, goal):

    if start == goal:
        return [start]

    stack = [start]

    # Stores only parent information
    parent = {start: None}

    while stack:

        current = stack.pop()

        if current == goal:
            return reconstruct_path(parent, goal)

        for next_state in get_neighbors(current):

            if next_state not in parent:

                parent[next_state] = current
                stack.append(next_state)

    return None


# ==========================================
# MAIN
# ==========================================

print("========== 8-PUZZLE USING DFS ==========")

start = read_state("INITIAL")
goal = read_state("GOAL")

print("\nInitial State:")
display(start)

print("Goal State:")
display(goal)

if not is_solvable(start, goal):

    print("No solution exists for these two states.")

else:

    print("\nSearching using DFS...")

    solution = dfs(start, goal)

    if solution is None:

        print("No solution found.")

    else:

        print("\nSolution Found!")
        print("Number of moves:", len(solution) - 1)

        for step, state in enumerate(solution):
            print("Step", step)
            display(state)
```

```python
# 8-Puzzle using Iterative Deepening Search (IDS)
# 0 represents the blank space


def read_state(name):
    print(f"\nEnter {name} state:")

    state = []

    for i in range(9):

        while True:
            try:
                value = int(input(f"Enter element {i + 1}: "))

                if value < 0 or value > 8:
                    print("Enter a number from 0 to 8.")
                    continue

                if value in state:
                    print("That number is already entered.")
                    continue

                state.append(value)
                break

            except ValueError:
                print("Please enter an integer.")

    return tuple(state)


def display(state):
    print()

    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])

    print()


def get_neighbors(state):
    neighbors = []

    # Find blank position
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    # Up, Down, Left, Right
    moves = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        # Check whether move is valid
        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            # Move tile into blank
            new_state[blank], new_state[new_blank] = (
                new_state[new_blank],
                new_state[blank]
            )

            neighbors.append(tuple(new_state))

    return neighbors


def inversion_count(state):
    values = [x for x in state if x != 0]

    count = 0

    for i in range(len(values)):
        for j in range(i + 1, len(values)):

            if values[i] > values[j]:
                count += 1

    return count


def is_solvable(start, goal):
    return (
        inversion_count(start) % 2
        == inversion_count(goal) % 2
    )


# -------------------------------------------------
# Depth Limited DFS
# -------------------------------------------------

def depth_limited_search(state, goal, depth, path):

    # Goal found
    if state == goal:
        return path

    # Depth limit reached
    if depth == 0:
        return None

    # Explore children using DFS
    for next_state in get_neighbors(state):

        # Do not go back to a state already in
        # the current path
        if next_state not in path:

            path.append(next_state)

            result = depth_limited_search(
                next_state,
                goal,
                depth - 1,
                path
            )

            if result is not None:
                return result

            # Backtrack
            path.pop()

    return None


# -------------------------------------------------
# Iterative Deepening Search
# -------------------------------------------------

def ids(start, goal):

    depth = 0

    while True:

        print("Searching with depth limit:", depth)

        # Current DFS path
        path = [start]

        result = depth_limited_search(
            start,
            goal,
            depth,
            path
        )

        if result is not None:
            return result

        depth += 1


# -------------------------------------------------
# MAIN PROGRAM
# -------------------------------------------------

print("========================================")
print("       8-PUZZLE USING IDS")
print("========================================")


# Read initial state
start = read_state("INITIAL")


# Read goal state
goal = read_state("GOAL")


print("\nInitial State:")
display(start)

print("Goal State:")
display(goal)


# Check whether solution is possible
if not is_solvable(start, goal):

    print("These two states are not solvable.")
    print("No solution exists.")

else:

    print("\nPuzzle is solvable.")
    print("Starting IDS...\n")

    solution = ids(start, goal)

    print("\n========== SOLUTION FOUND ==========")
    print("Number of moves:", len(solution) - 1)

    for step, state in enumerate(solution):

        print("Step", step)
        display(state)
```
