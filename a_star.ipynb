```python

import heapq

# Misplaced tiles heuristic
def heuristic(state, goal):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# Generate possible next states
def neighbors(state):
    result = []
    i = state.index(0)
    r, c = i // 3, i % 3

    moves = [(-1, 0), (1, 0),
             (0, -1), (0, 1)]

    for dr, dc in moves:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            j = nr * 3 + nc
            new = list(state)

            new[i], new[j] = new[j], new[i]
            result.append(tuple(new))

    return result


# A* Search
def astar(start, goal):
    pq = [(heuristic(start, goal), 0, start)]
    g_cost = {start: 0}
    parent = {start: None}

    while pq:
        f, g, state = heapq.heappop(pq)

        if g != g_cost[state]:
            continue

        if state == goal:
            path = []

            while state is not None:
                path.append(state)
                state = parent[state]

            return path[::-1]

        for next_state in neighbors(state):
            new_g = g + 1

            if new_g < g_cost.get(next_state, float('inf')):
                g_cost[next_state] = new_g
                parent[next_state] = state

                h = heuristic(next_state, goal)
                heapq.heappush(
                    pq, (new_g + h, new_g, next_state)
                )

    return None


# Display the 3x3 board
def display(state):
    for i in range(0, 9, 3):
        print(*state[i:i+3])
    print()


# Input
print("Enter initial state (0 for blank):")
start = tuple(map(int, input().split()))

print("Enter goal state:")
goal = tuple(map(int, input().split()))

if sorted(start) != list(range(9)) or sorted(goal) != list(range(9)):
    print("Invalid input! Use numbers 0 to 8 exactly once.")
else:
    path = astar(start, goal)

    if path:
        print("\nSolution found!")
        print("Number of moves:", len(path) - 1)

        for step, state in enumerate(path):
            print("Step", step)
            display(state)
    else:
        print("No solution exists.")
```
