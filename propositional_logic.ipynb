```python

from sympy import symbols, sympify
from sympy.logic.inference import satisfiable

# Input propositional symbols
names = input("Enter symbols separated by spaces: ").split()
variables = {name: symbols(name) for name in names}

# Input knowledge base
n = int(input("Enter number of statements in KB: "))
kb = []

for i in range(n):
    statement = input(f"Enter statement {i+1}: ")
    kb.append(sympify(statement, locals=variables))

# Input query
query = sympify(input("Enter query: "), locals=variables)

# Combine all KB statements using AND
knowledge_base = kb[0]
for statement in kb[1:]:
    knowledge_base = knowledge_base & statement

# Check KB AND NOT(query)
result = satisfiable(knowledge_base & ~query)

print("\nKnowledge Base:", knowledge_base)
print("Query:", query)

if result is False:
    print("Result: KB entails the query")
else:
    print("Result: KB does NOT entail the query")
```
