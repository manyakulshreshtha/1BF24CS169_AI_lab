```python

import random

# Count attacking queen pairs
def heuristic(state):
    h = 0
    n = len(state)

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or \
               abs(state[i] - state[j]) == abs(i - j):
                h += 1

    return h


# Find best neighboring state
def best_neighbor(state):
    n = len(state)
    best = state[:]
    best_h = heuristic(state)

    for col in range(n):
        for row in range(n):
            if row == state[col]:
                continue

            neighbor = state[:]
            neighbor[col] = row

            h = heuristic(neighbor)

            if h < best_h:
                best = neighbor
                best_h = h

    return best, best_h


# Hill Climbing
def hill_climbing(state):
    print("\nInitial state:", state)
    print("Initial heuristic:", heuristic(state))

    step = 0

    while True:
        h = heuristic(state)

        if h == 0:
            print("\nGoal reached!")
            return state

        next_state, next_h = best_neighbor(state)

        if next_h >= h:
            print("\nStuck at local minimum!")
            return None

        state = next_state
        step += 1

        print("Step", step, ":", state, "h =", next_h)


# Display board
def display(state):
    n = len(state)

    for row in range(n):
        for col in range(n):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


# Main program
n = int(input("Enter number of queens: "))

state = list(map(int, input(
    f"Enter {n} row positions (0 to {n-1}): "
).split()))

if n < 1 or len(state) != n or any(r < 0 or r >= n for r in state):
    print("Invalid initial state!")
else:
    print("\nInitial board:")
    display(state)

    solution = hill_climbing(state)

    if solution is not None:
        print("\nFinal solution:")
        display(solution)
    else:
        print("\nTry a different initial state.")

```
