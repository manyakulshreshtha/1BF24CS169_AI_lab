```python

import random
import math

# Count attacking queen pairs
def heuristic(state):
    h = 0

    for i in range(8):
        for j in range(i + 1, 8):
            if state[i] == state[j] or \
               abs(state[i] - state[j]) == abs(i - j):
                h += 1

    return h


# Generate a random neighboring state
def random_neighbor(state):
    neighbor = state[:]

    col = random.randint(0, 7)
    row = random.randint(0, 7)

    while row == neighbor[col]:
        row = random.randint(0, 7)

    neighbor[col] = row

    return neighbor


# Simulated Annealing
def simulated_annealing(state):
    temperature = 100
    cooling_rate = 0.999
    step = 0

    while temperature > 0.001:
        current_h = heuristic(state)

        if current_h == 0:
            return state

        next_state = random_neighbor(state)
        next_h = heuristic(next_state)

        delta = next_h - current_h

        if delta <= 0:
            state = next_state
        else:
            probability = math.exp(-delta / temperature)

            if random.random() < probability:
                state = next_state

        step += 1

        if step <= 20 or step % 1000 == 0:
            print("Step:", step,
                  "Heuristic:", heuristic(state),
                  "Temperature:", round(temperature, 3))

        temperature *= cooling_rate

    if heuristic(state) == 0:
        return state

    return None


# Display chessboard
def display(state):
    for row in range(8):
        for col in range(8):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


# Main program
print("Enter 8 row positions (0 to 7):")
state = list(map(int, input().split()))

if len(state) != 8 or any(r < 0 or r > 7 for r in state):
    print("Invalid initial state!")
else:
    print("\nInitial board:")
    display(state)

    print("\nInitial heuristic:", heuristic(state))

    solution = simulated_annealing(state)

    if solution is not None:
        print("\nSolution found!")
        print("Final state:", solution)
        print("Final heuristic:", heuristic(solution))
        print("\nFinal chessboard:")
        display(solution)
    else:
        print("\nNo solution found.")
        print("Try again or change temperature.")
```
